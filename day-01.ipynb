Functions: NumPy mean() | pandas mean()
Data: marks of 5 students = [70, 80, 90, 60, 85]


```python
import numpy as np
import pandas as pd
```

```python
marks = np.array([70, 80, 90, 60, 85])
print(marks)
```

```python
result = np.mean(marks)
print("numpy mean: ", result)
```

```python
marks_series = pd.Series(marks)
result1 = marks_series.mean()
print(result1)
```
