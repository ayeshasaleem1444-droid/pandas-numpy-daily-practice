Functions: NumPy max() | pandas max()
Data: my study hours  = [9, 8, 0, 7, 6, 8, 5, 4]


```python
import numpy as np
import pandas as pd
```

```python
study_hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
print(study_hours)
```

```python
result = np.max(study_hours)
print(result)
```

```python
hours_series = pd.Series(study_hours)
presult = hours_series.max()
print(presult)
```
