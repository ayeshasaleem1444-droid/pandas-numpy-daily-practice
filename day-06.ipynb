2026-11-21, cumsum()(cumulative sum) data = : [9, 8, 0, 7, 6, 8, 5, 4]
cumsum(): cumsum returns a whole array, not one number; each value is the total so far; the last value is the full sum.

```python
import pandas as pd
import numpy as np
```

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
result = np.cumsum(hours)
print(result)
```

```python
hours_series = pd.Series(hours)
result = hours_series.cumsum()
print(result)
```
