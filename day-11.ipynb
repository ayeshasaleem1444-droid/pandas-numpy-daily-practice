Day 11 — 2026-11-28 Functions: loc/iloc Data = [9, 8, 0, 7, 6, 8, 5, 4]

```python
import pandas as pd
import numpy as np
```

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
result = hours[1:4]
print(result)
hours[2:5]
```

```python
hours_series = pd.Series(hours)
result = hours_series.iloc[1:4]
result1 = hours_series.iloc[2:7]
print(result)
print(result1)
```

```python
result2 = hours_series.loc[1:4]
result3 = hours_series.iloc[2:7]
print(result2)
print(result3)
```
