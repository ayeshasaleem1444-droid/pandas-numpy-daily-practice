Functions Numpy: min() | Pandas: min() && argmin() / idxmin()
Data: [9, 8, 0, 7, 6, 8, 5, 4]

```python
import numpy as np
import pandas as pd
```

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
print(hours)
```

```python
result = np.min(hours)
print(result)
```

```python
hours_series = pd.Series(hours)
result = hours_series.min()
print(result)
```

```python
result2 = np.argmin(hours)
print(result2)
```

```python
result2 = hours_series.idxmin()
print(result2)
```
