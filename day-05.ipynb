2026-11-20, std()(standard deviation) data = : [9, 8, 0, 7, 6, 8, 5, 4]

```python
import pandas as pd
import numpy as np
```

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
result = np.std(hours)
print(result)
```

```python
hours_series = pd.Series(hours)
result = hours_series.std()
print(result)
```

ddof stands for Delta Degrees of Freedom — plain English: how much to subtract from the count before dividing.

```python
result2 = np.std(hours, ddof=1) 
print(result2)
```
