Day 9 — 2026-11-25 Functions:Boolean masking | filtering Data = : [9, 8, 0, 7, 6, 8, 5, 4]

```python
import pandas as pd
import numpy as np
```

First move: ask a yes/no question of every value at once(in numpy)

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
mask = hours > 6
print(mask)
```

Second move: use the mask as a sieve,Only the values sitting under a True survive.(in numpy)

```python
hours[mask]
```

First move: ask a yes/no question of every value at once(in pandas)

```python
hours_series = pd.Series(hours)
mask = hours > 6
print(mask)
```

Second move: use the mask as a sieve,Only the values sitting under a True survive.(in pandas)

```python
hours_series[mask]
```
