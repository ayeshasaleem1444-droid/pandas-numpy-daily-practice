Day 8 — 2026-11-23 Functions: NumPy .sort() | pandas .sort_values() Data = : [9, 8, 0, 7, 6, 8, 5, 4]

```python
import pandas as pd
import numpy as np
```

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
print(f"hours before sorting: ",hours)
result = np.sort(hours)
print(f"hours after sorting: ",result)
```

```python
hours_series = pd.Series(hours)
result = hours_series.sort_values()
print(result)
```
