```python
import numpy as np
import pandas as pd
```

```python
study_hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
result = np.median(study_hours)
print(result)
```

```python
hours_series = pd.Series(study_hours)
result = hours_series.median()
print(result)
```
