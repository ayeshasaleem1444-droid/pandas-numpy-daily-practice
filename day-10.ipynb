Day 10 — 2026-11-26 Functions: where , Data =  [9, 8, 0, 7, 6, 8, 5, 4]

```python
import pandas as pd
import numpy as np
```

```python
hours = np.array([9, 8, 0, 7, 6, 8, 5, 4])
result = np.where(hours >= 8, hours, 0)
print(result)
```

```python
np.where(hours >= 8, "strong", "light")
```

```python
np.where(hours >= 5 ,"good", "bad")
```

```python
hours_series = pd.Series(hours)
result = hours_series.where(hours_series >= 8,0)
print(result)
```
